```python
import pandas as pd
import numpy as np

# Read CSV
df = pd.read_csv("swtcp_mnist_iid_fr2_fr3_40pct_magz3_cosine_matrix_by_round.csv")

# Excel rows 2-41
data = df.iloc[0:40]

# Column B (rounds)
rounds = data.iloc[:, 1].tolist()

# Columns E:CZ
cosine_data = data.iloc[:, 4:104]

# Create empty DataFrames
cos_values = pd.DataFrame(index=cosine_data.index, columns=cosine_data.columns, dtype=float)
z_values = pd.DataFrame(index=cosine_data.index, columns=cosine_data.columns, dtype=float)

for col in cosine_data.columns:
    # Extract cosine values
    cos_values[col] = (
        cosine_data[col]
        .astype(str)
        .str.extract(r'cos=([-\d.]+)', expand=False)
        .astype(float)
    )

    # Extract z values
    z_values[col] = (
        cosine_data[col]
        .astype(str)
        .str.extract(r'z=([-\d.]+)', expand=False)
        .astype(float)
    )
```

```python
print("\nCosine values:")
print(cos_values)
```

```python
print("\nZ values:")
print(z_values)
```
