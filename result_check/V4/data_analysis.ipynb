```python
import pandas as pd
import numpy as np

# Read CSV
df = pd.read_csv("swtcp_mnist_iid_fr2_fr3_40pct_batchmad_cosine_matrix_by_round.csv")

# Excel rows 2-41
data = df.iloc[0:100]

# Column B (rounds)
rounds = data.iloc[:, 1].tolist()

# Columns E:CZ
cosine_data = data.iloc[:, 4:104]

# Create empty DataFrames
cos_values = pd.DataFrame(index=cosine_data.index, columns=cosine_data.columns, dtype=float)
z_values = pd.DataFrame(index=cosine_data.index, columns=cosine_data.columns, dtype=float)

for col in cosine_data.columns:
    # Extract cosine values
    cos_values[col] = (
        cosine_data[col]
        .astype(str)
        .str.extract(r'cos=([-\d.]+)', expand=False)
        .astype(float)
    )

    # Extract z values
    z_values[col] = (
        cosine_data[col]
        .astype(str)
        .str.extract(r'z=([-\d.]+)', expand=False)
        .astype(float)
    )
```

```python
print("\nCosine values:")
print(cos_values)
```

```python
import numpy as np
print("\nZ values:")
z = pd.DataFrame(z_values)
#print(z)z
z
```

```python
good = [[] for _ in range(len(z_values))]
bad = [[] for _ in range(len(z_values))]

for row_idx, row in z_values.iterrows():
    for col_name, value in row.items():
        if pd.notna(value):
            #print(col_name, value)
            client = col_name.replace("client_", "")
            client = int(client)
            if client in fr2_list:
                bad[row_idx].append(value)
            else:
                good[row_idx].append(value)
```

```python
good_z = pd.DataFrame(good)
good_z
```

```python
bad_z = pd.DataFrame(bad)
bad_z
```

```python
import pandas as pd

bad_z = pd.DataFrame(bad)
good_z = pd.DataFrame(good)

# Per row statistics
summary = pd.DataFrame({
    "good_low": good_z.min(axis=1),
    "good_high": good_z.max(axis=1),
    "good_avg": good_z.mean(axis=1),

    "bad_low": bad_z.min(axis=1),
    "bad_high": bad_z.max(axis=1),
    "bad_avg": bad_z.mean(axis=1)
})

#print(summary)
dx = pd.DataFrame(summary)
```
