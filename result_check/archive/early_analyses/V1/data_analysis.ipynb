```python
# Read CSV (header is kept as usual)
df = pd.read_csv("swtcp_mnist_iid_40pct_free_rider_client_list.csv")

# First column (A) and second column (B), starting from Excel row 2
col_a = df.iloc[:, 0]
col_b = df.iloc[:, 1]

# Create lists
fr2_list = col_b[col_a == "FR2"].tolist()
fr3_list = col_b[col_a == "FR3"].tolist()

print("FR2:", fr2_list)
print("FR2:", len(fr2_list))
print("FR3:", fr3_list)
print("FR3:", len(fr3_list))
```
